```python
# Para cargar el dataset de ELCA 2010 y mostrar las primeras filas del DataFrame sin el nombre de columnas. No está reconociendo el nombre de las columnas, por lo que asigna solo el índice de la columna.
UChoques_2010_raw = pd.read_csv( # Carga el archivo CSV sin encabezado
    "UChoques-csv.tab",
    sep="\t", # Especifica el separador de columnas como tabulación
    header=None, # Indica que no hay fila de encabezado en el archivo CSV
)

print(UChoques_2010_raw.to_string()) # Muestra las primeras 10 filas del DataFrame sin encabezado, mostrando todos los valores de las filas y columnas.
```

```python
# Para estandarizar y transformar las bases hay que primero organizar y nombrar la base, con el siguiente diccionario
bloques = [
    {"choque": 2,  "conteo": 3,  "respuestas": [4, 5, 6, 7]},
    {"choque": 8,  "conteo": 9,  "respuestas": [10, 11, 12]},
    {"choque": 13, "conteo": 14, "respuestas": [15, 16, 17, 18]},
    {"choque": 19, "conteo": 20, "respuestas": [21, 22]},
    {"choque": 23, "conteo": 24, "respuestas": [25, 26]},
    {"choque": 27, "conteo": 28, "respuestas": [29]},
    {"choque": 30, "conteo": 31, "respuestas": [32]},
]
bloques
```

```python
# Luego se construye con el diccionario de arriba la base de datos por hogar

import pandas as pd
registros = []
for _, fila in UChoques_2010_raw.iterrows():
    consecutivo = fila.iloc[1]
    for bloque in bloques:
        choque = fila.iloc[bloque["choque"]]
        if pd.isna(choque):
            continue
        conteo = fila.iloc[bloque["conteo"]]
        respuestas = []
        for col in bloque["respuestas"]:
            valor = fila.iloc[col]
            if pd.notna(valor):
                respuestas.append(str(valor))
        registros.append({
            "consecutivo": consecutivo,
            "choque": choque,
            "conteo": conteo,
            "respuesta": "; ".join(respuestas)
        })
Uchoques_2010_long = pd.DataFrame(registros)
Uchoques_2010_long
```

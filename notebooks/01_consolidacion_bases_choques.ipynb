## Dejo Uchoques en formato long

```python
import unicodedata
import re

def normalizar_nombre_choque(nombre):
    nfkd = unicodedata.normalize('NFKD', str(nombre))
    sin_acentos = ''.join(c for c in nfkd if not unicodedata.combining(c))
    return re.sub(r'[^a-z0-9]+', '_', sin_acentos.lower()).strip('_')

# UChoques
df_wide_u = Uchoques_2010_long.copy()
df_wide_u['choque_norm'] = df_wide_u['choque'].apply(normalizar_nombre_choque)

partes_u = []
for choque_n in sorted(df_wide_u['choque_norm'].unique()):
    subset = (
        df_wide_u[df_wide_u['choque_norm'] == choque_n]
        .set_index('consecutivo')[['conteo', 'respuesta_1', 'respuesta_2', 'respuesta_3', 'respuesta_4']]
    )
    subset.columns = [
        f'choque_{choque_n}' if c == 'conteo' else f'{c}_{choque_n}'
        for c in subset.columns
    ]
    partes_u.append(subset)

UChoques_2010 = pd.concat(partes_u, axis=1).reset_index()

# RChoques
df_wide_r = Uchoques_2010_long.copy()
df_wide_r['choque_norm'] = df_wide_r['choque'].apply(normalizar_nombre_choque)

partes_r = []
for choque_n in sorted(df_wide_r['choque_norm'].unique()):
    subset = (
        df_wide_r[df_wide_r['choque_norm'] == choque_n]
        .set_index('consecutivo')[['conteo', 'respuesta_1', 'respuesta_2', 'respuesta_3', 'respuesta_4']]
    )
    subset.columns = [
        f'choque_{choque_n}' if c == 'conteo' else f'{c}_{choque_n}'
        for c in subset.columns
    ]
    partes_r.append(subset)

RChoques_2010 = pd.concat(partes_r, axis=1).reset_index()

print(UChoques_2010.shape, RChoques_2010.shape)
UChoques_2010.head()
```

```python
UChoques_2010_conteo = (
    Uchoques_2010_long
    .assign(choque_norm=lambda df: df['choque'].apply(normalizar_nombre_choque))
    .pivot_table(index='consecutivo', columns='choque_norm', values='conteo', aggfunc='sum', fill_value=0)
    .reset_index()
)
UChoques_2010_conteo.columns.name = None
UChoques_2010_conteo.columns = (
    ['consecutivo'] + 
    [f'choque_{normalizar_nombre_choque(c)}' for c in UChoques_2010_conteo.columns[1:]]
)
UChoques_2010_conteo.columns.name = None

UChoques_2010_conteo
```

```python
UChoques_respuestas_2010_conteo = (
    Uchoques_2010_long[['consecutivo','respuesta_1','respuesta_2','respuesta_3','respuesta_4']]
    .melt(id_vars='consecutivo', value_name='respuesta')
    .dropna(subset=['respuesta'])
    [['consecutivo','respuesta']]
    .groupby(['consecutivo','respuesta'])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)
UChoques_respuestas_2010_conteo.columns.name = None
UChoques_respuestas_2010_conteo.columns = (
    ['consecutivo'] + 
    [f'resp_{normalizar_nombre_choque(c)}' for c in UChoques_respuestas_2010_conteo.columns[1:]]
)
UChoques_respuestas_2010_conteo
```

```python
print(set(Uchoques_2010_long['consecutivo']) - set(UChoques_respuestas_2010_conteo['consecutivo']))
print(set(UChoques_respuestas_2010_conteo['consecutivo']) - set(Uchoques_2010_long['consecutivo']))
```

```python
UChoques_2010 = UChoques_2010_conteo.merge(
    UChoques_respuestas_2010_conteo,
    on="consecutivo",
    how="inner"
)
UChoques_2010
```

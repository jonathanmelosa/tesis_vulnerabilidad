# 1. Cuántos hogares únicos presentes hay en las tres olas: 2010, 2013, 2016, (urbano/rural, por estratos, regiones, departamentos y municipios)

## Hogares presentes en las tres olas

```python
# Hogares únicos por ola (no es el valor real, ya que son solo únicos por ola, no presentes en todas las olas)
ola1 = set(hogar_elca_longitudinal_clean.query("ola == 1")["consecutivo"].unique()) # estaban en 2010 
ola2 = set(hogar_elca_longitudinal_clean.query("ola == 2")["consecutivo"].unique())
ola3 = set(hogar_elca_longitudinal_clean.query("ola == 3")["consecutivo"].unique())

print(f"Hogares en ola 1 (2010): {len(ola1):,}")
print(f"Hogares en ola 2 (2013): {len(ola2):,}")
print(f"Hogares en ola 3 (2016): {len(ola3):,}")
```

```python
#Hogares presentes en todas las olas, sin precencia en alguna ola es "0"
presencia = pd.crosstab(
    hogar_elca_longitudinal_clean["consecutivo"],
    hogar_elca_longitudinal_clean["ola"]
)
presencia["patron"] = (
    presencia[[1, 2, 3]]
    .astype(int)
    .astype(str)
    .agg("".join, axis=1)
)
presencia
```

```python
# Conteo de hogares por patrón
resumen = presencia["patron"].value_counts().to_frame(name="conteo")

# Agregar porcentaje
resumen["porcentaje"] = (
    resumen["conteo"] / resumen["conteo"].sum() * 100
).round(2)

# Agregar fila de total
resumen.loc["Total"] = {
    "conteo": resumen["conteo"].sum(),
    "porcentaje": 100.00
}
print(resumen.sort_values(by="conteo", ascending=False))
```

## urbano/rural

```python
hogar_elca_longitudinal_clean_presencia = hogar_elca_longitudinal_clean.merge(
    presencia,
    on="consecutivo",
    how="inner"
)
hogar_elca_longitudinal_clean_presencia
```

```python
print(f"Hogares en ola 1 (2010) que están también en al menos otra ola: {9853 - 773:,}")
```

```python
hogar_elca_longitudinal_clean_filtrada = hogar_elca_longitudinal_clean_presencia.query("patron != '100'") 
tabla = pd.pivot_table(
    hogar_elca_longitudinal_clean_filtrada,
    values="consecutivo",
    columns="ola",
    index="zona",
    aggfunc="count",
    margins=True,
    margins_name="Total"
)
porcentaje = (tabla.div(tabla.loc["Total"], axis=1) * 100).round(2)

resultado = tabla.astype(str) + " (" + porcentaje.astype(str) + "%)"

print(resultado)
```

# 3. Cuántos hogares se perdieron en 2013 y cuántos en 2016 (Composición de la pérdida: urbano/rural, por estratos departamentos y municipios)

### intento 2

```python
# Hogares únicos por ola
ola1 = set(hogar_elca_longitudinal_clean.query("ola == 1")["consecutivo"].unique()) # estaban en 2010 
ola2 = set(hogar_elca_longitudinal_clean.query("ola == 2")["consecutivo"].unique())
ola3 = set(hogar_elca_longitudinal_clean.query("ola == 3")["consecutivo"].unique())

# Attrition entre olas
perdidos_2013 = ola1 - ola2          # estaban en 2010 pero no en 2013
perdidos_2016 = ola2 - ola3          # estaban en 2013 pero no en 2016
perdidos_total = ola1 - ola3         # estaban en 2010 pero no llegaron a 2016
nunca_recuperados = ola1 - ola2 - ola3  # perdidos en 2013 que tampoco volvieron en 2016

print(f"Hogares en ola 1 (2010): {len(ola1):,}")
print(f"Hogares en ola 2 (2013): {len(ola2):,}")
print(f"Hogares en ola 3 (2016): {len(ola3):,}")
print()
print(f"Perdidos entre 2010→2013: {len(perdidos_2013):,} ({len(perdidos_2013)/len(ola1)*100:.1f}%)")
print(f"Perdidos entre 2013→2016: {len(perdidos_2016):,} ({len(perdidos_2016)/len(ola2)*100:.1f}%)")
print(f"Perdidos en total (2010→2016): {len(perdidos_total):,} ({len(perdidos_total)/len(ola1)*100:.1f}%)")
```

```python
# Composición del attrition por zona (urbano/rural)
df = hogar_elca_longitudinal_clean

# Hogares perdidos entre ola 1 y ola 2, con su zona de ola 1
df_ola1 = df.query("ola == 1")[["consecutivo", "zona"]].drop_duplicates("consecutivo")
df_ola1["perdido_2013"] = df_ola1["consecutivo"].isin(perdidos_2013)
df_ola1["perdido_total"] = df_ola1["consecutivo"].isin(perdidos_total)

print("Attrition 2010→2013 por zona:")
print(df_ola1.groupby("zona")["perdido_2013"].agg(["sum", "mean"])
      .rename(columns={"sum": "n_perdidos", "mean": "tasa"})
      .assign(tasa=lambda x: (x["tasa"]*100).round(1)))

print("\nAttrition 2010→2016 por zona:")
print(df_ola1.groupby("zona")["perdido_total"].agg(["sum", "mean"])
      .rename(columns={"sum": "n_perdidos", "mean": "tasa"})
      .assign(tasa=lambda x: (x["tasa"]*100).round(1)))
```

```python
# Attrition por región
print("Attrition 2010→2013 por región:")
print(df_ola1.merge(
    df.query("ola == 1")[["consecutivo", "region"]].drop_duplicates("consecutivo"),
    on="consecutivo"
).groupby("region")["perdido_2013"].agg(n_perdidos="sum", tasa="mean")
 .assign(tasa=lambda x: (x["tasa"]*100).round(1))
 .sort_values("tasa", ascending=False))
```
